```python
import random

def play_game():
    print("\n" + "=" * 45)
    print("        NUMBER GUESSING GAME")
    print("=" * 45)

    levels = {
        "1": ("Easy", 1, 50),
        "2": ("Medium", 1, 100),
        "3": ("Hard", 1, 500)
    }

    print("\nSelect Difficulty:")
    print("1. Easy (1-50)")
    print("2. Medium (1-100)")
    print("3. Hard (1-500)")

    while True:
        choice = input("\nEnter your choice (1/2/3): ")
        if choice in levels:
            level, start, end = levels[choice]
            break
        print("❌ Invalid choice. Please select 1, 2, or 3.")

    secret_number = random.randint(start, end)
    attempts = 0

    print(f"\nYou selected {level} mode.")
    print(f"Guess the number between {start} and {end}.")

    while True:
        try:
            guess = int(input("Enter your guess: "))
            attempts += 1

            if guess < start or guess > end:
                print(f"⚠️ Please enter a number between {start} and {end}.")
                continue

            if guess < secret_number:
                print("📉 Too Low!")
            elif guess > secret_number:
                print("📈 Too High!")
            else:
                print("\n🎉 Congratulations!")
                print(f"You guessed the correct number: {secret_number}")
                print(f"Total Attempts: {attempts}")
                break

        except ValueError:
            print("❌ Please enter a valid number.")

while True:
    play_game()

    again = input("\nDo you want to play again? (y/n): ").lower()
    if again != "y":
        print("\nThank you for playing! Goodbye. 👋")
        break
```
